```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder
```

```python
df=pd.read_csv("covid_toy.csv")
df.head()
```

```python
df["city"].value_counts()   
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(df.drop(columns=["has_covid"]),df["has_covid"],test_size=0.2)
```

```python
x_train.head()
```

```python
#adding simple imputer to fever column
si=SimpleImputer()
x_train_fever=si.fit_transform(x_train[["fever"]])


#also the test data
x_test_fever=si.transform(x_test[["fever"]])

x_train_fever.shape
```

```python
#ordinalencoden-> cough column
oe=OrdinalEncoder()
x_train_cough=oe.fit_transform(x_train[["cough"]])

# Also the test data
x_test_cough=oe.transform(x_test[['cough']])
x_train_cough.shape
```

```python
ohe = OneHotEncoder(drop='first', sparse_output=False)

x_train_city=ohe.fit_transform(x_train[["city",'gender']])
x_test_city=ohe.transform(x_test[["city",'gender']])
```

```python
#extracting age column



x_train_age=x_train.drop(columns=["fever","gender","city","cough"]).values
x_test_age=x_test.drop(columns=["fever","gender","city","cough"]).values
 
```

```python
x_train_transformed = np.concatenate([x_train_age, x_train_fever, x_train_cough, x_train_city], axis=1)
x_test_transformed=np.concatenate([x_test_age,x_test_fever,x_test_cough,x_test_city], axis=1)
x_train_transformed.shape
```

```python
#from using column transformer
from sklearn.compose import ColumnTransformer
transformer = ColumnTransformer(transformers=[
    ("tnf", SimpleImputer(), ["fever"]),
    ("tnc", OrdinalEncoder(categories=[["Mild", "Strong"]]), ["cough"]),
    ("tng", OneHotEncoder(sparse_output=False, drop='first'), ["gender", "city"])
], remainder="passthrough")
x_train_tran = transformer.fit_transform(x_train)
x_test_tran = transformer.transform(x_test)
x_train_tran.shape

```
